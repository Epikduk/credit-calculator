Чтобы воспользоваться кредитным калькулятором, вам необходимо ввести:

```python
property_value = service.validation_float('Введите стоимость недвижимости: ')
print(f'Введенная стоимость недвижимости: {property_value}')
```

- первоначальный взнос (руб.):

```python
initial_payment = service.validation_float('Введите первоначальный взнос: ')
print(f'Введенный первоначальный взнос: {initial_payment}')
```

```python
mortgage_term = service.validation_int('Введите срок ипотеки: ')
print(f'Введеный срок ипотеки: {mortgage_term}')
```
